```python
from langchain_huggingface import HuggingFacePipeline
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate

llm=HuggingFacePipeline.from_model_id(
    model_id='HuggingFaceTB/SmolLM-135M-Instruct',
    task='text-generation',
    pipeline_kwargs={
        'max_new_tokens': 256
    }
)
```

```python
algo_tutor_template="""
You are an expert computer science engineer. Explain the following algorithms: {algorithm_name}

Follow the exact structure:

### 1. The concept (What it is?)
Explain the problem this algorithm solves in 2 sentences.

### 2. The efficiency (Complexity analysis)
Define the time complexity and space complexity of the algorithm {algorithm_name}

### 3. Sample trace run
Show a simple step-by-step dry-run using the sample input: {sample_input}


Answer::
"""
```

```python
prompt=PromptTemplate(
    template=algo_tutor_template,
    input_variables=["algorithm_name", "sample_input"],
)
```

```python
final_prompt=prompt.format(algorithm_name="Bubble Sort", sample_input="[5,1,4,2]")
print(final_prompt)
```

```python
chain=prompt | llm

result=chain.invoke({
    "algorithm_name": "Binary Search",
    "sample_input": "Target 7 in list [1,3,5,7,9,11,13,15,16]"
})
```

```python
print(result)
```

```python
template=ChatPromptTemplate(
    [
        ("system", "You are a calculator that responds with reasoning about the calculation"),
        ("human", "Answer what is 10 multiplied by 20"),
        ("ai", "10*20=200 as its's equivalent to adding 10 to itself 20 times"),
        ("human", "{question}")
    ]
)
```

```python
chain=template | llm

response=chain.invoke({
    "question": "What is the square root of 10"
})
```

```python
print(response)
```
