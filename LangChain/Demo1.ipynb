```python
from langchain.agents import create_agent
from langchain_core.tools import tool
```

```python
@tool
def get_weather(location: str)->str:
    """Get weather in a given location"""
    return f"The weather in {location} is sunny and 25 Degree Celsius"
```

```python
agent=create_agent(
    model="groq:openai/gpt-oss-20b",
    tools=[get_weather],
    system_prompt="You are a weather agent. You are given a location and you need to return the weather of that location."
)
```

```python
result=agent.invoke({
    "messages": [
        {"role": "user", "content": "Whats the weather in Tokyo compared to London?"}
    ]
})
```
