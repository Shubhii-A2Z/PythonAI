```python
from groq import Groq
from dotenv import load_dotenv

load_dotenv()

client = Groq()
```

```python
Q = ("Roger has 5 tennis balls. He buys 2 more cans of tennis balls. "
     "Each can has 3 tennis balls. How many tennis balls does he have now?")

try1=client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {
            "role": "user",
            "content": f"Q: {Q}\nA:The answer (just the number) is"
        }
    ],
    max_completion_tokens=100,
    temperature=0
)

print(try1.choices[0].message.content)
```

```python
Q = ("Roger has 5 tennis balls. He buys 2 more cans of tennis balls. "
     "Each can has 3 tennis balls. How many tennis balls does he have now?")

try2=client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
        {
            "role": "user",
            "content": f"Q: {Q}\nA:The answer is"
        }
    ],
    temperature=0
)

print(try2.choices[0].message.content)
```
