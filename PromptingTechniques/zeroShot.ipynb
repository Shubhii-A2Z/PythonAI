# Zero-Shot Prompting with Groq

**Zero-Shot Prompting** means asking an LLM to perform a task directly without giving it any prior examples or training instances. You provide clear instructions and the input, relying entirely on the model's pre-trained knowledge to fulfill the request.

```python
from groq import Groq
from dotenv import load_dotenv

load_dotenv()

client=Groq()

system_prompt="""
    You are a helpful assistant. Summarize the user's text query and concisely in 1 sentence, preserving most important ideas.
"""

user_query="""
    Artificial Intelligence is changing how people work, learn and communicate. Modern AI system can analyze large amount of data,
    recognize patterns, generate text and images, and help people solve complex problems more quickly.
    However AI also raise concerns about privacy, inaccurate informations, job displacement, and unfair bias.
    Because of these opportunities and risks, organizations must use AI more responsibly, keep humans involved in important decisions, and regularly
    evaluate whether their systems are safe, accurate and fair.
"""

response=client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
        {
        "role": "system",
        "content": system_prompt # system instructions
        },
        {
            "role": "user",
            "content": user_query # input
        }
    ]
)

print(response.choices[0].message.content)
```
