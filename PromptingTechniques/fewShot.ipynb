```python
from groq import Groq
from dotenv import load_dotenv

load_dotenv()

client=Groq()

system_prompt="""
    You analyze sentiment and respond only in XML using this structure:
    <sentiment>
        <label>positive|negative|neutral</label>
        <confidence>0.0-1.0</confidence>
    </sentiment>
"""

examples=[
    (
        "I absolutely loved the new restaurant downtown!",
        """
        <sentiment>
            <label>positive</label>
            <confidence>0.95</confidence>
        </sentiment>
        """
    ),
    (
        "The movie was okay, nothing special.",
        """
        <sentiment>
            <label>neutral</label>
            <confidence>0.80</confidence>
        </sentiment>
        """
    ),
    (
        "Worst customer service I've ever experienced.",
        """
        <sentiment>
            <label>negative</label>
            <confidence>0.97</confidence>
        </sentiment>
        """
    )
]

user_query="The hotel room was clean and the staff were friendly, but the wifi barely worked."

few_shot_input="\n\n".join(
    f"Example input: {example_input}\n Example output: {example_output}"
    for example_input, example_output in examples
)

few_shot_input+=f"\n\nInput: {user_query}\nOutput:"

response=client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
        {
        "role": "system",
        "content": system_prompt # system instructions
        },
        {
            "role": "user",
            "content": user_query # input
        }
    ]
)

print(response.choices[0].message.content)
```
