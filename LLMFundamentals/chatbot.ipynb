```python
def stateless_gpt_demo(prompt: str)->str:
    from dotenv import load_dotenv
    from groq import Groq

    load_dotenv()

    client=Groq()
    response=client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    return response.choices[0].message.content
```

```python
stateless_gpt_demo("Who was the PM of UK before 2020")
```

```python
stateless_gpt_demo("Who was before him?")
```
