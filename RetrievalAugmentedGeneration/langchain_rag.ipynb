```python
from langchain_core.documents import Document
from pathlib import Path

# To Load and Parse PDF documents
from langchain_community.document_loaders import PyPDFLoader
```

## Creating a Langchain Document Abstract Class which stores any kind of document (text, pdf, csv, html etc)

```python
docs_by_hands=[
    Document(page_content="Python is a programming language", metadata={"source": "intro.txt", "level": "beginner"}),
    Document(page_content="Embeddings convert text into vectors", metadata={"source": "intro1.txt", "level": "beginner"}),
]

print([d.page_content for d in docs_by_hands])
```

```python
PDFS=Path("pdfs")

pdf_path=sorted(PDFS.glob("*.pdf"))[0]

print(pdf_path)
```

## Creating Loader Instance

```python
loader=PyPDFLoader(str(pdf_path))
```

## Loading the entire PDF in pages all at once

```python
pages=loader.load()
```

```python
for i, page in enumerate(pages):
    print(f"Page {i+1}")
    print(page.page_content)
    print("\n")

print(pages[0].metadata)
```

## Using lazy load to load one page at a time

```python
for i, page in enumerate(loader.lazy_load()):
    print(page.metadata)
    if i==1:
        break
```
